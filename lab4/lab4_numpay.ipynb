```python
import numpy as np
```

```python
# Для воспроизводимости результатов
```

```python
rng = np.random.default_rng(42)
```

```python
# Размерность
```

```python
n = 85
i = np.arange(n)
j = np.arange(n)
```

```python
# Матрица M: вне диагонали 2,
# на диагонали значения 3, 5, 7, ..., 171
```

```python
M = np.full((n, n), 2, dtype=np.int64)
np.fill_diagonal(M, 2 * i + 3)
```

```python
# Матрица N
```

```python
N = np.full((n, n), 4, dtype=np.int64)
```

```python
# Чётные строки: 5 в столбцах 2, 5, 8, ...
```

```python
N[::2, 2::3] = 5
```

```python
# Нечётные строки: 7 в нечётных столбцах
```

```python
N[1::2, 1::2] = 7
```

```python
# Вектор p: 1 + случайная величина N(2, 4)
# Дисперсия равна 4, стандартное отклонение равно 2
```

```python
p = 1 + rng.normal(loc=2, scale=2, size=n)
```

```python
# Вектор q — поэлементное произведение строк N
```

```python
q = N[24, :] * N[71, :]
```

```python
# Вычисление f
```

```python
# Индекс максимального квадрата элемента q
```

```python
k_star = np.argmax(q**2)
f = 4 * N[k_star, 59]
```

```python
# Вычисление g
```

```python
# Предполагается, что в произведении используются
# элементы M[i, j] * p[j], где:
# i = 24, ..., 69; j = 13, ..., 27.
```

```python
values = 7e-6 * M[24:70, 13:28] * p[13:28]
g = np.sum(np.prod(values**2, axis=1))
```

```python
print("Матрица M:")
print(M)

print("\nМатрица N:")
print(N)

print("\nВектор p:")
print(p)

print("\nВектор q:")
print(q)

print("\nk* =", k_star)
print("f =", f)
print("g =", g)
```
